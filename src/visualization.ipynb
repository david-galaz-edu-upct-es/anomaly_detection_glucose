```python
ts = pd.read_json(path_or_buf="../diabetes/personal-data/glucose_timeseries.json", typ= "series")
```

```python
fig, ax = plt.subplots()
ts.plot()

plt.axhline(mean, color = "red")
plt.axhline(mean + std, color = "hotpink")
plt.axhline(mean - std, color = "hotpink")
plt.axhline(mean + 2*std, color = "deeppink")
plt.axhline(mean - 2*std, color = "deeppink")
```
